```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day5/11866번: 요세푸스 문제 0/test.txt')
N, K = list(map(int, sys.stdin.readline().strip().split()))
circle = deque([i for i in range(1, N+1)])

pop_list = []
while circle:
    if K == 1:
        pop = circle.popleft()
        pop_list.append(pop)
        continue

    for i in range(K-1):
        pop = circle.popleft()
        circle.append(pop)
    
    pop = circle.popleft()
    pop_list.append(pop)

result = str(pop_list)
result = result.replace('[', '<').replace(']', '>')
print(result)
```
