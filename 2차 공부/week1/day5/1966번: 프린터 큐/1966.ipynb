```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day5/1966번: 프린터 큐/test.txt')
case_num = int(sys.stdin.readline().strip())

total_list = []
for idx, line in enumerate(sys.stdin.readlines()):
    if idx % 2 == 0:
        total_list.append(list(map(int, line.strip().split())))
    else:
        case_queue = deque()
        for idx_, num in enumerate(line.strip().split()):
            case_queue.append([idx_, int(num)])
        total_list.append(case_queue)

check = False
for i in range(0, 2*case_num, 2):
    n_number, find_num = total_list[i]
    queue_printer = total_list[i+1]
    count = 0
    while check == False:
        max_num = max(queue_printer, key = lambda x:x[1])[1]
        out = queue_printer[0][1]
        if max_num == out:
            idx__, num = queue_printer.popleft()
            count += 1
            if idx__ == find_num:
                print(count)
                check = True
        else:
            pop = queue_printer.popleft()
            queue_printer.append(pop)
    check = False
```
