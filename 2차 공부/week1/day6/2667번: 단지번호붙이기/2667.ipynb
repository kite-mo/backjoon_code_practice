BFS Ver.

```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day6/2667번: 단지번호붙이기/test.txt')
input = sys.stdin.readline

N = int(input())
maze = []

for _ in range(N):
    maze.append(list(map(int, input().strip())))

x = [0, 0, 1, -1]
y = [1, -1, 0, 0]

counts_dict = {}

for i in range(N):
    for j in range(N):

        if maze[i][j] == 0:
            continue
        
        queue = deque()
        start_x, start_y = i, j
        queue.append([start_x, start_y])
        maze[start_x][start_y] = 0
        counts = 1

        while queue:
            pop_x, pop_y = queue.popleft()
            for k in range(4):
                move_x, move_y = pop_x + x[k], pop_y + y[k]
                if 0 <= move_x < N and 0 <= move_y < N and maze[move_x][move_y] == 1:
                    queue.append([move_x, move_y])
                    maze[move_x][move_y] = 0
                    counts += 1

        counts_dict[(start_x, start_y)] = counts

sorted_values = sorted(counts_dict.values())

print(len(sorted_values))
for val in sorted_values:
    print(val)
```

DFS Ver.

```python
import sys
sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day6/2667번: 단지번호붙이기/test.txt')
input = sys.stdin.readline

N = int(input().strip())
maze = [list(map(int, input().strip())) for _ in range(N)]

dx = [0, 0, 1, -1]
dy = [1, -1, 0, 0]

def dfs(x, y):
    maze[x][y] = 0
    cnt = 1
    for k in range(4):
        nx, ny = x + dx[k], y + dy[k]
        if 0 <= nx < N and 0 <= ny < N and maze[nx][ny] == 1:
            cnt += dfs(nx, ny)
    return cnt

sizes = []
for i in range(N):
    for j in range(N):
        if maze[i][j] == 1:
            sizes.append(dfs(i, j))

sizes.sort()
print(len(sizes))
print(*sizes, sep="\n")
```
