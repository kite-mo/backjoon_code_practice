```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day6/1021번: 회전하는 큐/test.txt')
input = sys.stdin.readline
N, M = map(int, input().split())
num_list = list(map(int, input().split()))
N_queue = deque(range(1, N+1))

check_dict = {val : idx for idx, val in enumerate(N_queue)}

change_counts = 0
for num in num_list:     

    while num != N_queue[0]:
        
        length = len(N_queue)
        position = check_dict[num]

        if position <= length//2:
            # method 2
            pop_num = N_queue.popleft()
            N_queue.append(pop_num)
            change_counts += 1

        else:
            # method 3
            pop_num = N_queue.pop()
            N_queue.insert(0, pop_num)
            change_counts += 1

    # method 1
    if num == N_queue[0]:
        pop_num = N_queue.popleft()
        check_dict = {val : idx for idx, val in enumerate(N_queue)}

print(change_counts)
```
