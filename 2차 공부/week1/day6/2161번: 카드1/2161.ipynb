```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day6/2161번: 카드1/test.txt')
input = sys.stdin.readline
N = int(input().strip())
card_queue = deque(range(1, N+1))

pop_cards = []

while len(card_queue) != 1:

    pop_card = card_queue.popleft()
    pop_cards.append(pop_card)

    move_card = card_queue.popleft()
    card_queue.append(move_card)

pop_cards = pop_cards + list(card_queue)

print(*pop_cards, sep=' ')
```
