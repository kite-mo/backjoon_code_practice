```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day7/7569번: 토마토/test.txt')
input = sys.stdin.readline
M, N, H = map(int, input().split())

tomato_box = []
for h in range(H):
    tomato_box.append([])
    for n in range(N):
        tomato_box[h].append(list(map(int, input().split())))

move_x = [0, 0, 1, -1, 0, 0] 
move_y = [1, -1, 0, 0, 0, 0]
move_z = [0, 0, 0, 0, 1, -1]

raw_tomato_counts = 0
initial_position = deque()
for h in range(H):
    for n in range(N):
        for m in range(M):
            tomato = tomato_box[h][n][m]
            if tomato == 1:
                initial_position.append([h, n, m])
            elif tomato == 0:
                raw_tomato_counts += 1 # 기존 안 익은 토마토 개수 카운트

if raw_tomato_counts == 0:
    print(0)
else:   
    day_counts = -1
    while initial_position:
        day_counts += 1
        for iter_num in range(len(initial_position)):
            z, y, x = initial_position.popleft()
            for i in range(6):
                dx, dy, dz = x + move_x[i], y + move_y[i], z + move_z[i]
                if 0 <= dx < M and 0 <= dy < N and 0 <= dz < H and tomato_box[dz][dy][dx] == 0:
                    initial_position.append([dz, dy, dx])
                    tomato_box[dz][dy][dx] = 1
                    raw_tomato_counts -= 1 # 안 익은 토마토에서 익히는 토마토 개수 빼기

    if raw_tomato_counts > 0:
        print(-1)
    else:
        print(day_counts)
```
