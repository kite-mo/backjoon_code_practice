```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day4/2178번: 미로 탐색/test.txt')
N, M = map(int, sys.stdin.readline().split())
graph = [list(map(int, line.strip())) for line in sys.stdin.readlines()]
visited = [[False] * M for _ in range(N)]

queue = deque()
start_x, start_y = 0,0
queue.append([start_x, start_y])

move_x = [-1, 1, 0, 0] # 상, 하
move_y = [0, 0, -1, 1] # 좌, 우

while queue:
    x, y = queue.popleft()
    visited[x][y] = True

    if x == N-1 and y == M-1:
        print(graph[x][y])

    for i in range(4):
        new_x, new_y = x + move_x[i], y + move_y[i]
        if 0<=new_x<N and 0<=new_y<M and graph[new_x][new_y] == 1 and visited[new_x][new_y] == False:
            queue.append([new_x, new_y])
            graph[new_x][new_y] = graph[x][y] + 1
            visited[new_x][new_y] = True
```
