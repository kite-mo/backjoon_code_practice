```python
import sys
from collections import deque

sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day4/2606번: 바이러스/test.txt')
node_num = int(sys.stdin.readline().strip())
edge_num = int(sys.stdin.readline().strip())

graph = {}
check_list = [False]
for i in range(1, node_num + 1):
    graph[i] = []
    check_list.append(False)

for line in sys.stdin.readlines():
    node_1, node_2 = list(map(int, line.strip().split()))
    graph[node_1].append(node_2)
    graph[node_2].append(node_1)

start_node = 1
queue = deque()
queue.append(start_node)
check_list[start_node] = True

while queue:
    pop = queue.popleft()
    for node in graph[pop]:
        if not check_list[node]:
            queue.append(node)
            check_list[node] = True

print(sum(check_list[2:]))
```
