```python
import sys
```

```python
sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day3/DFS_BFS_예시문제/음료수_얼려_먹기/test.txt')
N, M = map(int, sys.stdin.readline().rstrip().split())

graph = []
for s in sys.stdin.readlines():
    graph.append(list(map(int, s.rstrip())))
```

```python
graph
```

```python
def dfs(n, m):
    if n<0 or m<0 or n>=N or m>=M:
        return False
    if graph[n][m] == 0:
        graph[n][m] = 1
        dfs(n, m-1)
        dfs(n+1, m)
        dfs(n-1, m)
        dfs(n, m+1)
        return True
    else:
        return False

counts = 0
for n in range(N):
    for m in range(M):
        if dfs(n, m) == True:
            print(n, m)
            counts += 1

print(counts)
```
