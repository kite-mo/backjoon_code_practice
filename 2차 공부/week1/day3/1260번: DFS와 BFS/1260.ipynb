```python
import sys, os
from collections import deque   
```

```python
sys.stdin = open('/Users/kitemo/코딩테스트/git/backjoon_code_practice/2차 공부/8월/week1/day3/1260번: DFS와 BFS/test.txt')
node_num, line_num, start_node = [int(i) for i in sys.stdin.readline().rstrip().split()]

check_node = [False] * (node_num + 1)
graph = [[] for i in range(node_num + 1)]

for line in sys.stdin.readlines():
    node, pair_node = map(int, line.split())
    graph[node].append(pair_node)
    graph[pair_node].append(node)

graph = [sorted(v) for v in graph]
```

```python
def dfs(graph, check_node, start_node, result):
    check_node[start_node] = True
    result.append(start_node)
    print(start_node, end = ' ')
    pair_nodes = graph[start_node]
    for pair_node in pair_nodes:
        if not check_node[pair_node]:
            dfs(graph, check_node, pair_node, result)
    return result

def bfs(graph, check_node, start_node):
    result = []
    queue = deque()
    queue.append(start_node)
    check_node[start_node] = True

    while queue:
        pop_node = queue.popleft()
        print(pop_node, end = ' ')
        result.append(pop_node)
        for pair_node in graph[pop_node]:
            if not check_node[pair_node]:
                queue.append(pair_node)
                check_node[pair_node] = True
    return result

dfs_result = dfs(graph, check_node, start_node, [])
check_node = [False] * (node_num + 1)
print('')
bfs_result = bfs(graph, check_node, start_node)
```
